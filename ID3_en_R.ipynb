```r
# Définir les probabilités
p_soleil <- 5/9
p_nuages <- 4/9

p_soleil_oui <- 2/5
p_soleil_non <- 3/5

p_nuages_oui <- 1/4
p_nuages_non <- 3/4

# Calculer les termes d'entropie
entropie_soleil <- -p_soleil_oui * log2(p_soleil_oui) - p_soleil_non * log2(p_soleil_non)
entropie_nuages <- -p_nuages_oui * log2(p_nuages_oui) - p_nuages_non * log2(p_nuages_non)

# Calculer l'entropie totale
entropie_meteo <- p_soleil * entropie_soleil + p_nuages * entropie_nuages

# Afficher les résultats
print(paste("entropie_soleil:", entropie_soleil))
print(paste("entropie_nuages:", entropie_nuages))
print(paste("entropie_meteo:", entropie_meteo))

```

```r
# Définir les probabilités
p_presents <- 4/9
p_absents <- 5/9

p_presents_oui <- 2/4
p_presents_non <- 2/4

p_absents_oui <- 1/5
p_absents_non <- 4/5

# Calculer les termes d'entropie
entropie_presents <- - p_presents_oui * log2(p_presents_oui) - p_presents_non * log2(p_presents_non)
entropie_absents <- - p_absents_oui * log2(p_absents_oui) - p_absents_non * log2(p_absents_non)

# Calculer l'entropie totale
entropie_amis <- p_presents * entropie_presents + p_absents * entropie_absents


# Afficher les résultats
print(paste("entropie_present:", entropie_presents))
print(paste("entropie_absents:", entropie_absents))
print(paste("entropie_amis:", entropie_amis))



```

```r
# Définir les probabilités
p_fort <- 4/9
p_faible <- 5/9

p_fort_oui <- 0/4
p_fort_non <- 4/4

p_faible_oui <- 3/5
p_faible_non <- 2/5

# Calculer les termes d'entropie
entropie_fort <-  - p_fort_non * log2(p_fort_non)
entropie_faible <- - p_faible_oui * log2(p_faible_oui) - p_faible_non * log2(p_faible_non)

# Calculer l'entropie totale
entropie_vent <- p_fort * entropie_fort + p_faible * entropie_faible

# Afficher les résultats
entropie_fort
entropie_faible
entropie_vent
```
